## 1. Preprocessing

- Notch filter at 50Hz, filter 0.1 - 100Hz
- Create a copy, and filter it by 1 - 100Hz for ICA (requirement for MNE-ICALabel)
- After ICA, filter 0.1 - 30Hz
- Create epochs, create evokeds and write them to a file for each participant

```python
# Creates a lot of output!!!
# Create all evoked files for all subjects
for subject_id in subject_ids:
    raw = preprocess_subject(bids_root, str(subject_id))
    
    run_ica(raw)

    # Create Evokeds and write them into a file
    create_evokeds(raw, subject_id)
```
